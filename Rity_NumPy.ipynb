```python
# Numpy (numerical python)
# is a open source python library widely used in data science / machine learning / gen AI.
# it provides powerful N dimensional array data structure.
# most efficient and fastest library in python
# why it is fast - underlying is C pre - compiled code

# numpy is ndarray object - which represents a homogenous N dimentional array of items
```

```python
# numpy is a module which needs to be imported
# pip install numpy
import numpy as np
```

```python
# 1 D array
a = np.array([2, 4, 6, 8, 10])
```

```python
print(a)
print(a[2])
```

```python
# 2 D array
b = np.array([[1, 11, 111, 1111], [21, 22, 23, 24], [31, 32, 33, 34]])
print(b)
```

```python
# accessing a numpy array element
print(b[1, 3])
```

```python
# modifying an element i.e numpy arrays are mutable
b[1, 3] = 200
print(b)
```

```python
# slicing
# same python slicing (i.e list, string) works with numpy
# there is key diference between python slicing and numpy slicing
# python list slicing -  returns a copy
# numpy slicing - returns a view of the original data

a = np.array([10, 20, 30, 40, 50])
print(a[:3])
```

```python
b = a[:3]
print(b)
```

```python
b[0] = 1000
print(b)
```

```python
print(a)
```

```python
# Numpy Attributes
# ndim - number of dimensions
# shape - returns a tuple / size of the array along each dimension
# size - total number of element in the array
# dtype - data type of numpy array

a = np.array([1, 2, 3, 4])
print(f"Dimension (ndim): {a.ndim}")
print(f"Shape : {a.shape}")
print(f"Size : {a.size}")
print(f"Data type : {a.dtype}")
```

```python
a = np.array([[1, 2, 3, 4], [21, 22, 23, 24], [31, 32, 33, 34], [41, 42, 43, 44]])
print(f"Dimension (ndim): {a.ndim}")
print(f"Shape : {a.shape}")
print(f"Size : {a.size}")
print(f"Data type : {a.dtype}")
```

```python
# Numpy has inbuilt function to create a array with initial values
a = np.zeros(2)
print(a)
```

```python
a = np.ones(4)
print(a)
```

```python
a = np.empty(4)
print(a)
```

```python
# arange - creates an aray with a range of element
a = np.arange(2, 9, 2)
print(a)
```

```python
# sorting and concatination
a = np.array([210, 20, 3, 410, 150])
print(np.sort(a))
```

```python
a = np.array([210, 20, 3, 410, 150])
b = np.array([1210, 120, 13, 1410, 1150])
print(np.concatenate((a, b)))
```

```python
# conditional indexing / vectorizaion example
b = np.array([[1, 11, 111, 1111], [21, 22, 23, 24], [31, 32, 33, 34]])
print(b[b > 100])
print(b[b % 2 == 0])
print(b[(b > 25) & (b < 100)])
```

```python
# copy
a = np.array([210, 20, 3, 410, 150])
a2 = a.copy()
print(a2)
a2[2] = 3000
print(a2)
print(a)
```

```python
# Array Operation 
a = np.array([1, 2, 3, 4, 5])
b = np.array([1, 2, 3, 4, 5])
print(a + b)
```

```python
print(a * b)
```

```python
# Broadcasting
# it allows numpy to [perform operation on arrays with different shapes by virtually expanding dimensions

a = np.array([[1, 2, 3, 4, 5],
             [11, 12, 13, 14, 15]])

#b = np.array([1, 2, 3, 4, 5])
b = np.array([10])

print(a.shape)
print(b.shape)

print(a + b)
```
