```python
# pandas series 
# series is one dimensional (array) capable of holding any data type (integers, string, floats, python object)
# it has an index, which lables each element

s1 = pd.Series([1, 2, 3, 4, 5, 6])
print(s1)
print(s1[0])
```

```python
# add custom index
s1 = pd.Series([1, 2, 3, 4, 5, 6], index=["a", 'b', 'c', 'd', 'e', 'f'])
print(s1)
```

```python
# series from dictionary
series_dict = {'apple' : 10, 'banana': 3, 'blueberry': 6}
s2 = pd.Series(series_dict)
print(s2)
```

```python
# Accessing a element in pandas series
print(s2['apple'])
```

```python
# slicing a series
print(s1[1:4])
```

```python
# vectorised operations
print(s1 + 2)
```

```python
# vectorised operations
print(s1 * 2)
```

```python
# filtering
print(s1[s1 > 2])
```

```python
# Data Frame
# 2 Dimnetsional size mutable tabular data structure 

# dictinary of list
data = {
    'Name' : ['Alice', 'Bob', 'Charlie', 'David', 'Alice1', 'Bob1', 'Charlie1', 'David1'],
    'Age' : [20, 22, 24, 26, 20, 22, 24, 26],
    'City' : ['NYC', 'Delhi', 'Houston', 'Canada', 'NYC', 'Delhi', 'Houston', 'Canada'],
    'Score' : [90, 91, 92, 93, 90, 91, 92, 93]
}

df = pd.DataFrame(data)
print(df)
```

```python
df = pd.DataFrame(data, index=['Student_1', 'Student_2', 'Student_3', 'Student_4'])
print(df)
```

```python
# Data loading / saving
df.to_csv('student_data.csv')
print("student.csv file created")
```

```python
# reading from csv files
file_read_csv = pd.read_csv('student_data.csv')
print(file_read_csv)
```

```python
df_missing = pd.DataFrame({
    'A' : [1, 2, np.nan, 4],
    'B' : [1, 2, 31, 4],
    'C' : [1, np.nan, np.nan, 4],
    'D' : [np.nan, 2, 22, 4]
})

print(df_missing)
```

```python
# group by
# menthod used to group data by some criteria then apply a sort of any function 

sales_data = {
    'Region' : ['East', 'West', 'East', 'West', 'East', 'North'],
    'Product:' : ['A', 'B', 'A', 'C', 'B', 'A'],
    'Sales' : [100, 101, 120, 300, 80, 8],
    'Units': [10, 11, 4, 1, 10, 8]
}

df_sales_data = pd.DataFrame(sales_data)
print(df_sales_data)
```

```python
print(df_sales_data.groupby('Region')['Sales'].sum())
```

```python
print(df_sales_data.groupby('Region')['Sales'].mean())
```

```python
print(df_sales_data.groupby('Region')['Units'].sum())
```
